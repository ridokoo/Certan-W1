# 1. Analisis korelasi
 Korelasi positif (+) → jika variabel pertama meningkat, variabel kedua cenderung meningkat.
Korelasi negatif (-) → jika variabel pertama meningkat, variabel kedua cenderung menurun.
Mendekati 0 → hubungan linear lemah.
Mendekati +1 atau -1 → hubungan linear kuat.

```python
import pandas as pd

path = "C:\\Users\\MSII\\Downloads\\shopping_trends.csv"

df = pd.read_csv(path)

previous_purchases = df["Previous Purchases"]

print("=== Statistik Previous Purchases ===")
print("Mean :", previous_purchases.mean())
print("Std  :", previous_purchases.std())
print("Min  :", previous_purchases.min())
print("Q1   :", previous_purchases.quantile(0.25))
print("Q2   :", previous_purchases.quantile(0.50))
print("Q3   :", previous_purchases.quantile(0.75))
print("Max  :", previous_purchases.max())
```

```python
df[['Age', 'Purchase Amount (USD)']].corr()
```

# 1. Korelasi Age vs Purchase Amount (USD)
Age vs Purchase Amount (USD) = -0.010424

Analisis:

Age vs Purchase Amount (USD) (-0.010424), berarti terdapat hubungan negatif yang sangat lemah antara usia pelanggan dengan jumlah pembelian. Nilai korelasi yang sangat mendekati 0 menunjukkan bahwa perubahan usia pelanggan hampir tidak memiliki hubungan dengan jumlah Purchase Amount (USD).

Kesimpulan:

Semakin bertambah usia pelanggan, jumlah pembelian tidak mengalami perubahan yang berarti. Dengan demikian, Age dan Purchase Amount (USD) hampir tidak memiliki hubungan linear.
