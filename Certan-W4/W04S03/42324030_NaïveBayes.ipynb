# Tugas dan praktikum Naive Bayes

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]
# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

# Bagian 1: Import library dan memuat dataset

#### Di bagian awal, kita memanggil tiga alat yang akan dipakai sepanjang program. NumPy dipakai untuk mengolah angka dalam bentuk array, Pandas dipakai untuk membuat tabel data, dan load_iris adalah fungsi dari scikit-learn untuk mengambil dataset Iris yang sudah tersedia. Setelah iris = load_iris() dijalankan, variabel iris berisi seluruh data: 150 bunga iris, masing-masing punya empat ukuran (panjang dan lebar kelopak, panjang dan lebar mahkota), beserta jenis spesiesnya.

```python
y
```

# Bagian 2: Mengubah data menjadi DataFrame
Data mentah dari scikit-learn berbentuk array yang sulit dibaca, jadi kita ubah menjadi tabel dengan pd.DataFrame, lengkap dengan nama kolomnya. Kemudian kita tambahkan kolom Species yang berisi nama spesies (setosa, versicolor, virginica), bukan lagi angka 0, 1, dan 2. Setelah tabel siap, kita pisahkan dua hal. X adalah semua kolom ukuran bunga yang menjadi bahan bagi model untuk belajar, sedangkan y adalah kolom Species yang menjadi jawaban yang ingin ditebak oleh model. Perintah print(X) hanya untuk melihat isi tabel ukurannya.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

# Bagian 3: Mengubah label teks menjadi angka
Model machine learning tidak bisa memproses tulisan seperti "setosa", jadi nama spesies perlu diganti dengan angka. LabelEncoder mengerjakan ini secara otomatis dengan mengurutkan nama spesies sesuai abjad, sehingga setosa menjadi 0, versicolor menjadi 1, dan virginica menjadi 2. Objek lab sengaja kita simpan karena nanti dibutuhkan untuk menerjemahkan angka hasil prediksi kembali menjadi nama spesies.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

# Bagian 4: Membagi data latih dan data uji
Sebelum model dilatih, data dibagi dua. Sebanyak 80 persen (120 bunga) dipakai untuk melatih model, dan 20 persen sisanya (30 bunga) disimpan untuk menguji seberapa pintar model tersebut. Pembagian ini penting karena kita ingin menguji model dengan data yang belum pernah dilihatnya, seperti ujian yang soalnya berbeda dari latihan. Pengaturan random_state=0 membuat pengacakan datanya selalu sama setiap kali program dijalankan, sehingga hasilnya konsisten. Pembagian dilakukan dua kali, satu untuk ukuran bunga dan satu untuk labelnya, dan keduanya tetap cocok karena memakai pengaturan acak yang sama.

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
result = NB_model.predict(X_test)
result
```

# Bagian 5: Melatih model Gaussian Naïve Bayes
Di sini model dibuat dan dilatih. Kita memilih Gaussian Naïve Bayes karena semua ukuran bunga berupa angka kontinu yang bisa dianggap menyebar mengikuti kurva normal. Saat fit dijalankan, model mempelajari data latih dengan cara menghitung rata-rata dan sebaran ukuran bunga untuk setiap spesies. Misalnya, model belajar bahwa bunga setosa umumnya punya mahkota yang sangat kecil, sedangkan virginica punya mahkota yang besar. Model juga mencatat seberapa sering tiap spesies muncul di data latih.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

# Bagian 6: Menguji model dan menghitung akurasi
Setelah belajar, model diminta menebak spesies dari 30 bunga di data uji lewat predict. Untuk tiap bunga, model membandingkan ukuran bunga itu dengan pola yang sudah dipelajari, lalu memilih spesies yang paling masuk akal. Hasil tebakan kemudian dibandingkan dengan jawaban yang sebenarnya memakai accuracy_score. Hasilnya 96,67 persen, artinya dari 30 bunga, 29 ditebak dengan benar dan hanya 1 yang salah.

```python
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```
